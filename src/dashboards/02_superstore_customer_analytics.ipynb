```python
%sql
WITH base AS (
  SELECT
    SUM(total_sales) AS revenue,
    SUM(total_profit) AS profit,
    SUM(total_orders) AS total_orders,
    COUNT(DISTINCT customer_id) AS customers
  FROM superstore_catalog.dev_semantic_layer.customer_kpi
)

SELECT
  -- Revenue
  CONCAT('$', ROUND(revenue/1000000,2), 'M') AS revenue,

  -- Profit
  CONCAT('$', ROUND(profit/1000,2), 'K') AS profit,

  -- Avg Order Value (correct calculation)
  CONCAT(
    '$',
    ROUND(revenue / NULLIF(total_orders, 0), 2)
  ) AS avg_order_value,

  -- Customers
  customers

FROM base;
```

```python
%sql
-- Top Customers Driving Revenue
SELECT
  customer_name AS `Customer`,
  region AS `Region`,
  segment AS `Segment`,

  CONCAT('$', ROUND(total_sales/1000000,2), 'M') AS `Sales`,

  CONCAT(
    CASE WHEN total_profit >= 0 THEN '▲ ' ELSE '▼ ' END,
    '$', ROUND(ABS(total_profit)/1000,2), 'K'
  ) AS `Profit`

FROM superstore_catalog.dev_semantic_layer.customer_kpi
WHERE overall_sales_rank <= 10
ORDER BY total_sales DESC;
```

```python
%sql
-- Customer Recency (Retention)
SELECT
  CASE
    WHEN recency_days <= 30 THEN 'Active'
    WHEN recency_days <= 90 THEN 'Warm'
    ELSE 'Churn Risk'
  END AS customer_status,
  COUNT(*) AS customers,
  CONCAT('$', ROUND(SUM(total_sales)/1000000,2), 'M') AS revenue

FROM superstore_catalog.dev_semantic_layer.customer_kpi
GROUP BY customer_status;
```
